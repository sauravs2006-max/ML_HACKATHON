# 11 · Folium maps: GPS, road network, reference route, predicted route (student-test)

```python
import sys, os
from pathlib import Path
_root = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src" / "pathsnap").is_dir()), None)
if _root is None:
    raise RuntimeError("src/pathsnap not found: unzip all zips of the set into the same pathsnap/ folder")
sys.path.insert(0, str(_root / "src")); os.chdir(_root / "notebooks")
from pathsnap import io, pipeline, viz

model, cfg = pipeline.load_artifacts("../artifacts")
os.makedirs("../outputs/maps", exist_ok=True)
```

```python
for tid in io.list_trips("student_test"):
    tr = io.load_trip("student_test", tid)
    out = pipeline.predict(tr, model, cfg["hmm"])
    pred = out["hmm"][out["hmm"] >= 0]
    viz.trip_map(tr, pred=pred, ref=tr["route"]).save(f"../outputs/maps/student_test_{tid}.html")
    print(tid, pipeline.evaluate(tr, out)["hmm"])
```
